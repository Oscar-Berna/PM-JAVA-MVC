```python
# ============================================================
# BLOQUE 2 - CONDICIONALES
# ============================================================
print("========== BLOQUE 2: CONDICIONALES ==========")

# ── 2.1 Estado del personaje segun % de vida ──
print("\n===== ESTADO DEL PERSONAJE =====")

vida     = 25.0
vida_max = 100.0
pct      = (vida / vida_max) * 100   # calcula el porcentaje

if pct <= 0:
    estado = 'MUERTO'
elif pct <= 25:
    estado = 'CRITICO'
elif pct <= 50:
    estado = 'HERIDO'
elif pct <= 75:
    estado = 'ESTABLE'
else:
    estado = 'SALUDABLE'

print(f'Vida: {vida:.0f}/{vida_max:.0f} ({pct:.0f}%) -> Estado: {estado}')

# ── 2.2 Tipo de ataque segun clase (match) ──
print("\n===== TIPO DE ATAQUE =====")

clase_personaje = 'Mago'
nivel_habilidad = 3

match clase_personaje:
    case 'Guerrero': tipo_ataque = 'Espada'
    case 'Mago':     tipo_ataque = 'Hechizo'
    case 'Arquero':  tipo_ataque = 'Flecha'
    case _:          tipo_ataque = 'Puno'

puede_usar_magia = (clase_personaje == 'Mago' and nivel_habilidad >= 3)

if puede_usar_magia:
    print(f'Clase: {clase_personaje} -> Bola de fuego!')
else:
    print(f'Clase: {clase_personaje} -> {tipo_ataque} basico')

# ── Ejercicio 2A - Logica de combate completa ──
print("\n===== EJERCICIO 2A: COMBATE =====")

vida_enemigo    = 40
ataque_jugador  = 35
nivel_jugador   = 6

bonificacion  = 10 if nivel_jugador >= 5 else 0
dano_total    = ataque_jugador + bonificacion
vida_restante = vida_enemigo - dano_total

print(f'Ataque base:   {ataque_jugador}')
print(f'Bonificacion:  {bonificacion}')
print(f'Dano total:    {dano_total}')
print(f'Vida restante: {vida_restante}')

if vida_restante <= 0:
    print('Resultado: Enemigo derrotado! +50 XP')
elif vida_restante <= 20:
    print('Resultado: Enemigo en estado critico')
else:
    print(f'Resultado: Enemigo resiste. Vida restante: {vida_restante}')
```

```python
# ============================================================
# BLOQUE 3 - CICLOS
# ============================================================
print("========== BLOQUE 3: CICLOS ==========")

# ── 3.1 for - Recorrer el inventario ──
print("\n===== INVENTARIO =====")

inventario = [
    'Espada de hierro',
    'Pocion de vida',
    'Escudo de madera',
    'Llave dorada'
]

# for con indice
print("-- Lista numerada --")
for i, item in enumerate(inventario, 1):
    print(f'{i}. {item}')

# Buscar item especifico
buscar = 'Pocion de vida'
if buscar in inventario:
    print(f'[OK] {buscar} encontrada')
else:
    print(f'[X]  {buscar} no disponible')

# ── 3.2 while - Combate por rondas ──
print("\n===== COMBATE POR RONDAS =====")

vida_hero    = 80
vida_enemigo = 60
ronda        = 1

while vida_hero > 0 and vida_enemigo > 0:
    vida_enemigo -= 15   # heroe ataca
    vida_hero    -= 10   # enemigo contraataca
    print(f'Ronda {ronda}: Hero={vida_hero} | Enemigo={vida_enemigo}')
    ronda += 1

print('VICTORIA!' if vida_hero > 0 else 'DERROTA')

# ── Ejercicio 3A - Acumular XP y subir nivel ──
print("\n===== SUBIR NIVEL =====")

xp           = 0
nivel        = 1
xp_necesario = 100
batallas     = [20, 35, 15, 40, 30]

for xp_ganado in batallas:
    xp += xp_ganado
    print(f'XP acumulado: {xp}')
    if xp >= xp_necesario:
        nivel += 1
        xp -= xp_necesario
        print(f'*** Subiste al Nivel {nivel}! XP restante: {xp} ***')
```
